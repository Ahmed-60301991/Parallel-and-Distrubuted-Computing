# Results Visualisations

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data for the original and enhanced explorers
original_moves = [1279, 1279, 1279, 1279]  # Number of moves in the original explorer
enhanced_moves = [127, 127, 127, 127]  # Number of moves in the enhanced explorer

original_time = [0.00, 0.00, 0.00, 0.00]  # Time taken by the original explorer
enhanced_time = [0.00122, 0.00111, 0.00127, 0.00125]  # Time taken by the enhanced explorer

original_moves_per_second = [1025523.77, 1005909.40, 1049191.24, 1152667.56]  # Avg moves per second for original
enhanced_moves_per_second = [76161.94, 77910.87, 82279.36, 93402.88]  # Avg moves per second for enhanced
```

## 1. Horizontal Bar Chart for Number of Moves with Enhanced at the Bottom

```python
plt.figure(figsize=(8, 5))

# Bar positions and height
y = np.arange(2)
moves = [np.mean(enhanced_moves), np.mean(original_moves)]  # Enhanced at the bottom
explorers = ['Enhanced', 'Original']  # Labels in the order we want

# Create horizontal bar chart
fig, ax = plt.subplots()
bars = ax.barh(y, moves, align='center', color=['red', 'blue'])

ax.set_xlabel('Number of Moves')
ax.set_title('Number of Moves Comparison')
ax.set_yticks(y)
ax.set_yticklabels(explorers)

# Adding data labels on the bars
for bar in bars:
    width = bar.get_width()
    ax.text(width + 10, bar.get_y() + bar.get_height() / 2,
            f'{width:.0f}', va='center', ha='left', fontsize=10)

plt.tight_layout()
plt.show()
```

## 2. Bar Chart for Average Moves per Second

```python
plt.figure(figsize=(8, 5))
fig, ax = plt.subplots()

# Bar width
width = 0.35  # Adjust bar width

# x-axis positions for each bar
x = np.arange(1)  # Only one position for both bars
rects1 = ax.bar(x - width/2, [np.mean(original_moves_per_second)], width, label='Original')
rects2 = ax.bar(x + width/2, [np.mean(enhanced_moves_per_second)], width, label='Enhanced')

# Labeling and formatting
ax.set_ylabel('Average Moves per Second')
ax.set_title('Average Moves per Second Comparison')
ax.set_xticks(x)
ax.set_xticklabels(['Original vs Enhanced'])
ax.legend()

# Function to label bars
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)

plt.tight_layout()
plt.show()
```

## 3. Line Chart for Execution Time Comparison

```python
plt.figure(figsize=(8, 5))
plt.plot(range(4), original_time, label='Original Explorer', marker='o', color='blue')
plt.plot(range(4), enhanced_time, label='Enhanced Explorer', marker='o', color='red')

plt.xlabel('Explorer Version')
plt.ylabel('Time Taken (s)')
plt.title('Execution Time Comparison')
plt.xticks(range(4), ['Run 1', 'Run 2', 'Run 3', 'Run 4'])
plt.legend()
plt.tight_layout()
plt.show()
```
